* This is a simple program that formats an event deadline.
* Standards: The program is expected to use at least one date/time feature and should run without syntax errors.
  * The program makes use of the `datetime` module to construct the datetime object via `datetime.date()` and format it into human-readable outputs with `.strftime()`.

```python
# We need the datetime module to perform date/time functions, so let's import it.
import datetime

event_title = input("Enter event title: ")
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

# Process the inputs and format them.
# It still makes use of some string manipulation functions besides datetime to format it properly.
# .strftime() formats the datetime object into various human-legible strings.
# A little information: %B represents month, %Y represents year, %d represents the day, and %A represents the weekday.
formatted_event = event_title.strip().upper()
date = datetime.date(year, month, day)
formatted_date = date.strftime("%B %d, %Y").replace(" 0", " ")
weekday = date.strftime("%A")

print("\nFormatted Output:")
print("Event:", formatted_event)
print("Date:", formatted_date)
print("Year:", year)
print("Month:", date.strftime("%B"))
print("Day:", date.strftime("%d"))
print("Weekday:", weekday)
```
